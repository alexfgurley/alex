```python
#  Prime Star Systems
arr_mag = np.array([
[2, 3, 5],
[14, 6, 8],
[11, 13, 17],
[7, 10, 13]
])

def sort_prime(array) :
    prime_arr = ([])
    prime = False
    
    for row in array :
        
        for x in row :
            prime = True
            
            if x <= 1 : 
                prime = False
                
            for i in range(2, x) : 
                if x % i == 0 :
                    prime = False
                    break 
                    
        for x in row :
            
            if prime == True : 
                prime_arr.append(row)     
                break
            
    return np.array(prime_arr)

print(sort_prime(arr_mag)) 
```

```python
# 5.2
def board_fill(num):
    array = []
    for row_idx in range(num):
        start = 1 if row_idx % 2 == 0 else 0 # if we want it to start with a 1 we just change row_idx % 2 == 0 instead of 1 
        row = [(start + col_idx) % 2 for col_idx in range(num)]
        array.append(row)
    return array

print(board_fill(8))
        
def reverse_board_fill(num):
    array = []
    for row_idx in range(num):
        start = 0 if row_idx % 2 == 0 else 1
        row = [(start + col_idx) % 2 for col_idx in range(num)]
        array.append(row)
    return array 

print(reverse_board_fill(8))
```

```python
# 6 Expanding Universe
def expand(array, num) : 
    x = 0
    expanded_array = []
    for word in array: 
        expanded_word = []
        n = len(word) 
        
        for i in range(n) : 
            expanded_word.append(word[i])
            
            if i != n-1 : 
                expanded_word.append(" " * num)
                
        expanded_array.append("".join(expanded_word))
    return expanded_array 

universe = np.array(['Galaxy', 'clusters'])
print(expand(universe, 2))
            
```

```python
# 7. Second-Brightest Star
def second_brightest(array) : 

    second_brightest_values = []
    
    for row in array : 
        
        brightest_value = 0
        second_brightest_value = 0 
        
        for num in row : 
            
            if num > brightest_value : 
                second_brightest_value = brightest_value
                brightest_value = num 
                
            elif num > second_brightest_value :
                second_brightest_value = num
                
        second_brightest_values.append(int(second_brightest_value))
        
    return f"The second brightest values are {second_brightest_values}."

stars = np.array([[1123, 1456, 1789, 1324, 1876],
[1567, 1987, 1678, 1405, 1589],
[1345, 1654, 1523, 1109, 1923],
[1298, 1890, 1367, 1784, 1432],
[1823, 1756, 1489, 1672, 1550]])

print(second_brightest(stars)) 
```
