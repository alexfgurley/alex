```python
# 3. Seaborn 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
sns.load_dataset("car_crashes")
ccdf = sns.load_dataset("car_crashes") 
ccdf = ccdf.dropna(axis=1) 
```

```python
# 6 
# What is the shape of the data set? 
print(ccdf.shape) # data set has 51 rows and 8 columns, and its overall shape is an 8 by 51 matrix

# Print first and last five entries 
print(f" {ccdf.head()} \n {ccdf.tail()} ") 

# List each column name and what it represents 
# total --> represents the total number of fatal car crashes per 1 billion miles in each state
# speeding --> percentage of drivers involved in fatal car crashes who were speeding in each state
# alcohol --> percentage of drivers involved in fatal car crashes who were alcohol impaired in each state 
# not_distracted --> percentage in each state that were not impaired 
# no_previous --> percentage of drivers with no prior accidents
# ins_premium --> car insurance premiums for each state
# ins_losses --> the losses incurred by insurance companies for insured driver
# all columns relate to fatal accidents 
# abbrev --> the abbraviation of each state that the row data corresponds to 
```

```python
# 7. Subplots 

fig, axes = plt.subplots(2, 2) 

axes[0, 0].scatter(ccdf["alcohol"], ccdf["not_distracted"], color="red") 
axes[0, 0].set_title("Alcohol vs not distracted accidents") 
axes[0,0].set_xlabel("alcohol related")
axes[0,0].set_ylabel("non distracted accidents")
axes[0,0].legend(title="alcohol vs not distracted", loc="best")

axes[0, 1].scatter(ccdf["total"], ccdf["speeding"], color="blue") 
axes[0, 1].set_title("Total vs speeding related accidents")
axes[0,1].set_xlabel("total accidents")
axes[0,1].set_ylabel("speeding related")

sns.histplot(ccdf["total"], ax=axes[1,0], color="green")
axes[1,0].set_title("total car accidents per state")
axes[1,0].set_xlabel("state")
axes[1,0].set_ylabel("number of fatal accidents")

sns.boxplot(ccdf["ins_losses"], ax=axes[1,1], color="black")
axes[1,1].set_title("insurance losses box plot")
axes[1,1].set_ylabel("insurance losses")

plt.tight_layout()
plt.show()

# 9. 
# subplot 1 shows the alcohol related car accidents vs the number of non-distracted driving incidents, wanted to show how alcohol relates to distracted driving.
# subplot 2 shows total accidents per state vs. speeding-related accidents. The goal was to see how many accidents were related to speed. 
# subplot 3 shows a histogram of the total car accidents per state to see which one had the most vs least. 
# subplot 4: the median insurance losses per state to see what the average expected insurance losses would be.

# What we've learned from this data set is that the majority of states have between 10 and 15 car crashes per 1 billion miles. Additionally, the average insurance losses per state are around 125-150 dollars per driver
```
